### 1.2.1 Deductive Reasoning

Deductive reasoning moves from **general principles to specific conclusions**. In AI, this means starting with a knowledge base of facts and rules, then using systematic methods (logic, search) to make inferences.

**Example:**
- **General Rule**: All canine animals have four legs
- **Specific Fact**: All dogs are canines
- **Deduction**: Therefore, dogs have four legs

**Core Characteristics:**
- Starts with established hypotheses and domain knowledge
- Uses logic-based systems (propositional logic, first-order logic)
- Highly interpretable - you can explain why a conclusion was reached
- Requires extensive domain knowledge from human experts
- Provably correct within the knowledge base

**Strengths:**
- Mathematically rigorous and provably correct
- Transparent decision-making process
- Works well in well-defined, narrow domains
- Requires less data than learning-based approaches

**Limitations:**
- Combinatorial explosion in large state spaces
- Difficulty encoding intuitive, non-explicit knowledge
- Requires significant human expertise to build knowledge bases
- Brittleness - fails on cases not anticipated by knowledge base

````python
# Example: Deductive Reasoning System
# Simple forward chaining inference engine

class DeductiveSystem:
    """
    A simple forward-chaining inference engine for propositional logic.
    Demonstrates deductive reasoning from a knowledge base.
    """
    
    def __init__(self):
        self.facts = set()  # Known facts
        self.rules = []     # Inference rules (premise -> conclusion)
    
    def add_fact(self, fact):
        """Add a known fact to the knowledge base."""
        self.facts.add(fact)
        print(f"Added fact: {fact}")
    
    def add_rule(self, premises, conclusion):
        """
        Add an inference rule.
        premises: list of required facts
        conclusion: fact that can be inferred
        """
        self.rules.append((premises, conclusion))
        print(f"Added rule: {' AND '.join(premises)} -> {conclusion}")
    
    def forward_chain(self):
        """
        Apply forward chaining to derive new facts.
        Continues until no new facts can be inferred.
        """
        inferred = True
        iterations = 0
        
        while inferred:
            inferred = False
            iterations += 1
            
            for premises, conclusion in self.rules:
                # Check if all premises are satisfied
                if all(p in self.facts for p in premises):
                    # Check if conclusion is not already known
                    if conclusion not in self.facts:
                        self.facts.add(conclusion)
                        print(f"Iteration {iterations}: Inferred '{conclusion}' from {premises}")
                        inferred = True
        
        return self.facts
    
    def query(self, fact):
        """Check if a fact can be proven from the knowledge base."""
        return fact in self.facts


# Example: Animal classification system
print("=== Deductive Reasoning: Animal Classification ==")
print()

system = DeductiveSystem()

# Add facts
system.add_fact("Tweety is a bird")
system.add_fact("Polly is a bird")
system.add_fact("Polly lives in cage")
print()

# Add rules
system.add_rule(["Tweety is a bird"], "Tweety can fly")
system.add_rule(["Polly is a bird"], "Polly can fly")
system.add_rule(["Polly is a bird", "Polly lives in cage"], "Polly is a pet")
system.add_rule(["Polly is a pet"], "Polly needs care")
print()

# Perform inference
print("Performing forward chaining...")
print()
derived_facts = system.forward_chain()

print()
print("=== All Known Facts ==")
for fact in sorted(derived_facts):
    print(f"  - {fact}")

print()
print("=== Query Examples ==")
queries = ["Tweety can fly", "Polly needs care", "Tweety is a pet"]
for q in queries:
    result = "TRUE" if system.query(q) else "FALSE"
    print(f"Query: '{q}' -> {result}")
````

### 1.2.2 Inductive Learning

Inductive learning moves from **specific examples to general patterns**. Instead of starting with rules, the system learns patterns from data.

**Example:**
- **Observation**: I saw a couple of dogs yesterday. Both had four legs.
- **Induction**: Therefore, all dogs have four legs.

Note that this is a form of potentially faulty logic based on generalizing specific experiences. However, combining it with statistical methods and large amounts of data can result in accurate (though not provably correct) inferences.

**Core Characteristics:**
- Learns from examples (data-driven)
- Discovers patterns without explicit programming
- Less interpretable ("black box" nature)
- Requires large amounts of data
- Example: Neural networks, decision trees, support vector machines

**Strengths:**
- Can capture intuitive, hard-to-articulate knowledge
- Scales with data and computational power
- Can discover novel patterns beyond human knowledge
- Works on unstructured data (images, audio, text)

**Limitations:**
- Requires massive amounts of data
- Computationally expensive to train
- Lack of interpretability (hard to explain decisions)
- Vulnerable to adversarial examples and biased data
- May learn spurious correlations

````python
# Example: Inductive Learning System
# Simple decision tree learner for classification

import numpy as np
from collections import Counter

class SimpleDecisionTree:
    """
    A basic decision tree classifier demonstrating inductive learning.
    Learns patterns from training examples rather than using explicit rules.
    """
    
    def __init__(self, max_depth=3):
        self.max_depth = max_depth
        self.tree = None
    
    def entropy(self, y):
        """Calculate entropy of labels."""
        counts = Counter(y)
        probs = [count / len(y) for count in counts.values()]
        return -sum(p * np.log2(p) for p in probs if p > 0)
    
    def information_gain(self, X, y, feature_idx, threshold):
        """Calculate information gain from a split."""
        parent_entropy = self.entropy(y)
        
        # Split data
        left_mask = X[:, feature_idx] <= threshold
        right_mask = ~left_mask
        
        if sum(left_mask) == 0 or sum(right_mask) == 0:
            return 0
        
        # Calculate weighted entropy
        n = len(y)
        left_entropy = self.entropy(y[left_mask])
        right_entropy = self.entropy(y[right_mask])
        
        child_entropy = (sum(left_mask) / n * left_entropy +
                        sum(right_mask) / n * right_entropy)
        
        return parent_entropy - child_entropy
    
    def find_best_split(self, X, y):
        """Find the best feature and threshold to split on."""
        best_gain = -1
        best_feature = None
        best_threshold = None
        
        for feature_idx in range(X.shape[1]):
            thresholds = np.unique(X[:, feature_idx])
            
            for threshold in thresholds:
                gain = self.information_gain(X, y, feature_idx, threshold)
                
                if gain > best_gain:
                    best_gain = gain
                    best_feature = feature_idx
                    best_threshold = threshold
        
        return best_feature, best_threshold
    
    def build_tree(self, X, y, depth=0):
        """Recursively build the decision tree."""
        # Stopping conditions
        if depth >= self.max_depth or len(np.unique(y)) == 1:
            return Counter(y).most_common(1)[0][0]
        
        feature_idx, threshold = self.find_best_split(X, y)
        
        if feature_idx is None:
            return Counter(y).most_common(1)[0][0]
        
        # Split data
        left_mask = X[:, feature_idx] <= threshold
        right_mask = ~left_mask
        
        # Build subtrees
        left_tree = self.build_tree(X[left_mask], y[left_mask], depth + 1)
        right_tree = self.build_tree(X[right_mask], y[right_mask], depth + 1)
        
        return {
            'feature': feature_idx,
            'threshold': threshold,
            'left': left_tree,
            'right': right_tree
        }
    
    def fit(self, X, y):
        """Train the decision tree on data."""
        self.tree = self.build_tree(X, y)
        return self
    
    def predict_sample(self, x, tree):
        """Predict class for a single sample."""
        if not isinstance(tree, dict):
            return tree
        
        if x[tree['feature']] <= tree['threshold']:
            return self.predict_sample(x, tree['left'])
        else:
            return self.predict_sample(x, tree['right'])
    
    def predict(self, X):
        """Predict classes for multiple samples."""
        return np.array([self.predict_sample(x, self.tree) for x in X])


# Example: Learn to classify animals
print("=== Inductive Learning: Animal Classification ==")
print()

# Training data: [has_feathers, can_fly, lays_eggs] -> class
X_train = np.array([
    [1, 1, 1],  # Bird
    [1, 1, 1],  # Bird
    [1, 0, 1],  # Flightless bird (e.g., penguin)
    [0, 0, 0],  # Mammal
    [0, 0, 0],  # Mammal
    [0, 0, 1],  # Reptile
    [0, 0, 1],  # Reptile
    [1, 1, 1],  # Bird
])

y_train = np.array(['bird', 'bird', 'bird', 'mammal', 'mammal', 'reptile', 'reptile', 'bird'])

# Train model
tree = SimpleDecisionTree(max_depth=3)
tree.fit(X_train, y_train)
print("Model trained on", len(X_train), "examples")
print()

# Test predictions
X_test = np.array([
    [1, 1, 1],  # Should predict bird
    [0, 0, 0],  # Should predict mammal
    [1, 0, 1],  # Should predict bird (flightless)
])

predictions = tree.predict(X_test)

test_descriptions = [
    "Has feathers, can fly, lays eggs",
    "No feathers, can't fly, doesn't lay eggs",
    "Has feathers, can't fly, lays eggs"
]

print("=== Test Predictions ==")
for desc, pred in zip(test_descriptions, predictions):
    print(f"{desc} -> {pred}")

print()
print("Note: The model learned these patterns from examples, not from explicit rules!")
````

1. **Simple Reflex Agent**: Acts based only on current percept

5. **Learning Agent**: Improves performance through experience

````python
# Example: Simple Agent Implementations

from abc import ABC, abstractmethod
from enum import Enum

class Environment:
    """Base class for environments."""
    
    @abstractmethod
    def get_percept(self):
        """Return current percept from environment."""
        pass
    
    @abstractmethod
    def execute_action(self, action):
        """Execute action and update environment."""
        pass


class Agent(ABC):
    """Base class for all agents."""
    
    def __init__(self):
        self.percept_history = []
    
    @abstractmethod
    def choose_action(self, percept):
        """Choose action based on current percept."""
        pass
    
    def perceive(self, percept):
        """Store percept in history."""
        self.percept_history.append(percept)
        return percept


# Example: Vacuum Cleaner Environment
class VacuumLocation(Enum):
    A = 0
    B = 1


class VacuumAction(Enum):
    SUCK = 0
    MOVE_LEFT = 1
    MOVE_RIGHT = 2


class VacuumEnvironment(Environment):
    """Two-location vacuum world."""
    
    def __init__(self):
        self.location = VacuumLocation.A
        self.dirty = {VacuumLocation.A: True, VacuumLocation.B: True}
        self.steps = 0
    
    def get_percept(self):
        """Return (location, is_dirty) tuple."""
        return (self.location, self.dirty[self.location])
    
    def execute_action(self, action):
        """Execute action and return reward."""
        self.steps += 1
        reward = 0
        
        if action == VacuumAction.SUCK:
            if self.dirty[self.location]:
                self.dirty[self.location] = False
                reward = 10  # Reward for cleaning
        elif action == VacuumAction.MOVE_LEFT:
            self.location = VacuumLocation.A
            reward = -1  # Small penalty for moving
        elif action == VacuumAction.MOVE_RIGHT:
            self.location = VacuumLocation.B
            reward = -1
        
        return reward
    
    def is_clean(self):
        """Check if environment is fully clean."""
        return not any(self.dirty.values())


# Simple Reflex Agent
class ReflexVacuumAgent(Agent):
    """
    Simple reflex agent: acts based only on current percept.
    """
    
    def choose_action(self, percept):
        location, is_dirty = percept
        
        if is_dirty:
            return VacuumAction.SUCK
        elif location == VacuumLocation.A:
            return VacuumAction.MOVE_RIGHT
        else:
            return VacuumAction.MOVE_LEFT


# Run simulation
print("=== Vacuum Cleaner Agent Simulation ==")
print()

env = VacuumEnvironment()
agent = ReflexVacuumAgent()

total_reward = 0
max_steps = 10

print("Initial state: A is dirty, B is dirty")
print()

for step in range(max_steps):
    percept = env.get_percept()
    agent.perceive(percept)
    action = agent.choose_action(percept)
    reward = env.execute_action(action)
    total_reward += reward
    
    loc_name = "A" if percept[0] == VacuumLocation.A else "B"
    dirty_status = "dirty" if percept[1] else "clean"
    print(f"Step {step + 1}: At {loc_name} ({dirty_status}) -> {action.name} (reward: {reward})")
    
    if env.is_clean():
        print()
        print(f"Environment clean after {step + 1} steps!")
        break

print()
print(f"Total reward: {total_reward}")
````

### 1.6.1 Supervised Learning

**Supervised learning** learns from labeled examples to predict outputs for new inputs.

**Key Idea**: Given training data with known answers (labels), learn a function that maps inputs to outputs.

**Mathematical Formulation**:
- Training data: \((X_1, y_1), (X_2, y_2), \ldots, (X_n, y_n)\)
- Goal: Learn function \(f\) such that \(y_i \approx f(X_i)\)

**Types of Supervised Learning**:

1. **Classification**: Predict categorical labels
   - Example: Email spam detection (spam/not spam)
   - Example: Image recognition (cat/dog/bird)

2. **Regression**: Predict numerical values
   - Example: House price prediction
   - Example: Temperature forecasting

**Analogy**: Learning with a teacher who provides correct answers

````python
# Minimal Example: Supervised Learning (Linear Regression)
import matplotlib.pyplot as plt

# Generate simple training data
np.random.seed(42)
X_train = np.array([1, 2, 3, 4, 5, 6, 7, 8])
y_train = 2 * X_train + 1 + np.random.randn(8) * 0.5  # y = 2x + 1 + noise

# Simple linear regression: find best fit line
# Using least squares: minimize sum of (y - (mx + b))^2
X_mean = np.mean(X_train)
y_mean = np.mean(y_train)
m = np.sum((X_train - X_mean) * (y_train - y_mean)) / np.sum((X_train - X_mean)**2)
b = y_mean - m * X_mean

print(f"Learned model: y = {m:.2f}x + {b:.2f}")
print(f"True model: y = 2.00x + 1.00")
print()

# Make predictions
X_test = np.array([9, 10])
y_pred = m * X_test + b

print("Predictions for new data:")
for x, y in zip(X_test, y_pred):
    print(f"  x = {x}: predicted y = {y:.2f}")

# Visualize
plt.figure(figsize=(10, 6))
plt.scatter(X_train, y_train, color='blue', s=100, label='Training data', zorder=3)
plt.scatter(X_test, y_pred, color='red', s=100, marker='s', label='Predictions', zorder=3)
X_line = np.linspace(0, 11, 100)
plt.plot(X_line, m * X_line + b, 'r--', label=f'Learned: y={m:.2f}x+{b:.2f}', linewidth=2)
plt.plot(X_line, 2 * X_line + 1, 'g:', label='True: y=2x+1', linewidth=2)
plt.xlabel('x', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.title('Supervised Learning: Linear Regression', fontsize=14, fontweight='bold')
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\nNote: Model learned from labeled training data!")
````

### 1.6.2 Unsupervised Learning

**Unsupervised learning** discovers patterns in data without labeled examples.

**Key Idea**: Find hidden structure in unlabeled data.

**Common Tasks**:

1. **Clustering**: Group similar data points
   - Example: Customer segmentation
   - Example: Document organization

2. **Dimensionality Reduction**: Compress data to fewer dimensions
   - Example: Visualizing high-dimensional data
   - Example: Feature extraction

3. **Anomaly Detection**: Find unusual patterns
   - Example: Fraud detection
   - Example: Network intrusion detection

**Analogy**: Learning without a teacher - discovering patterns on your own

````python
# Minimal Example: Unsupervised Learning (K-Means Clustering)

# Generate sample data with 3 natural clusters
np.random.seed(42)
cluster1 = np.random.randn(30, 2) + np.array([0, 0])
cluster2 = np.random.randn(30, 2) + np.array([5, 5])
cluster3 = np.random.randn(30, 2) + np.array([0, 5])
X = np.vstack([cluster1, cluster2, cluster3])

# Simple k-means clustering
k = 3
# Initialize centroids randomly
centroids = X[np.random.choice(len(X), k, replace=False)]

# Iterate to find clusters
for iteration in range(10):
    # Assign each point to nearest centroid
    distances = np.sqrt(((X[:, np.newaxis] - centroids)**2).sum(axis=2))
    labels = np.argmin(distances, axis=1)
    
    # Update centroids
    for i in range(k):
        if np.sum(labels == i) > 0:
            centroids[i] = X[labels == i].mean(axis=0)

print(f"Discovered {k} clusters in unlabeled data")
print("\nCluster centers:")
for i, center in enumerate(centroids):
    print(f"  Cluster {i+1}: ({center[0]:.2f}, {center[1]:.2f})")

# Visualize
plt.figure(figsize=(12, 5))

# Before clustering
plt.subplot(1, 2, 1)
plt.scatter(X[:, 0], X[:, 1], c='gray', s=50, alpha=0.6)
plt.title('Before Clustering\n(No labels)', fontsize=12, fontweight='bold')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.grid(True, alpha=0.3)

# After clustering
plt.subplot(1, 2, 2)
colors = ['red', 'blue', 'green']
for i in range(k):
    cluster_points = X[labels == i]
    plt.scatter(cluster_points[:, 0], cluster_points[:, 1], 
                c=colors[i], s=50, alpha=0.6, label=f'Cluster {i+1}')
plt.scatter(centroids[:, 0], centroids[:, 1], 
            c='black', s=200, marker='X', edgecolors='white', linewidths=2,
            label='Centroids', zorder=5)
plt.title('After Clustering\n(Patterns discovered)', fontsize=12, fontweight='bold')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\nNote: Clusters found WITHOUT any labels!")
````

### 1.6.3 Reinforcement Learning

**Reinforcement learning** learns by trial and error through interaction with an environment, receiving rewards or penalties.

**Key Idea**: An agent learns which actions to take by trying them and observing the results.

**Core Components**:
- **Agent**: The learner/decision maker
- **Environment**: What the agent interacts with
- **State**: Current situation
- **Action**: What the agent can do
- **Reward**: Feedback (positive or negative)

**Goal**: Maximize cumulative reward over time

**Examples**:
- Training a robot to walk (reward for staying upright)
- Game playing (reward for winning)
- AlphaZero learning chess through self-play

**Analogy**: Like training a dog - reward good behavior, penalize bad behavior

````python
# Minimal Example: Reinforcement Learning (Simple Bandit)

class SimpleBandit:
    """A simple multi-armed bandit problem."""
    
    def __init__(self, true_values):
        self.true_values = np.array(true_values)
        self.n_arms = len(true_values)
    
    def pull(self, arm):
        """Pull an arm and get reward."""
        # Reward is true value + noise
        return self.true_values[arm] + np.random.randn() * 0.1


class EpsilonGreedyAgent:
    """Agent using epsilon-greedy strategy."""
    
    def __init__(self, n_arms, epsilon=0.1):
        self.n_arms = n_arms
        self.epsilon = epsilon
        self.q_values = np.zeros(n_arms)  # Estimated values
        self.action_counts = np.zeros(n_arms)  # Times each arm pulled
    
    def choose_action(self):
        """Choose action using epsilon-greedy."""
        if np.random.random() < self.epsilon:
            return np.random.randint(self.n_arms)  # Explore
        else:
            return np.argmax(self.q_values)  # Exploit
    
    def update(self, action, reward):
        """Update estimates based on received reward."""
        self.action_counts[action] += 1
        # Incremental average
        self.q_values[action] += (reward - self.q_values[action]) / self.action_counts[action]


# Setup: 3 slot machines with different expected rewards
true_values = [1.0, 2.0, 1.5]  # Machine 2 is best!
bandit = SimpleBandit(true_values)
agent = EpsilonGreedyAgent(n_arms=3, epsilon=0.1)

# Learning process
n_steps = 1000
rewards_history = []
action_history = []

for step in range(n_steps):
    action = agent.choose_action()
    reward = bandit.pull(action)
    agent.update(action, reward)
    
    rewards_history.append(reward)
    action_history.append(action)

# Results
print("=== Reinforcement Learning: Multi-Armed Bandit ===")
print()
print("True values of slot machines:", true_values)
print("Learned estimates:", [f"{v:.2f}" for v in agent.q_values])
print()
print("Times each machine was selected:")
for i, count in enumerate(agent.action_counts):
    print(f"  Machine {i+1}: {int(count)} times ({count/n_steps*100:.1f}%)")
print()
print(f"Best machine: #{np.argmax(agent.q_values) + 1}")

# Visualize learning
plt.figure(figsize=(12, 5))

# Cumulative reward
plt.subplot(1, 2, 1)
cumulative_reward = np.cumsum(rewards_history)
plt.plot(cumulative_reward, linewidth=2)
plt.xlabel('Step', fontsize=12)
plt.ylabel('Cumulative Reward', fontsize=12)
plt.title('Learning Progress', fontsize=12, fontweight='bold')
plt.grid(True, alpha=0.3)

# Action selection over time
plt.subplot(1, 2, 2)
window = 100
action_proportions = []
for i in range(0, len(action_history), window):
    window_actions = action_history[i:i+window]
    proportions = [window_actions.count(a)/len(window_actions) for a in range(3)]
    action_proportions.append(proportions)

action_proportions = np.array(action_proportions)
x = np.arange(len(action_proportions)) * window
colors = ['red', 'green', 'blue']
for i in range(3):
    plt.plot(x, action_proportions[:, i], label=f'Machine {i+1}', 
             color=colors[i], linewidth=2)
plt.xlabel('Step', fontsize=12)
plt.ylabel('Selection Proportion', fontsize=12)
plt.title('Action Selection Over Time', fontsize=12, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\nNote: Agent learned through trial and error with rewards!")
````

### 1.7.1 Genetic Algorithms

**Genetic algorithms** apply evolutionary principles to optimization problems.

```{mermaid}
graph TD
    A[Initialize Population] --> B[Evaluate Fitness]
    B --> C[Selection]
    C --> D[Crossover]
    D --> E[Mutation]
    E --> B
    B --> F{Converged?}
    F -->|No| C
    F -->|Yes| G[Best Solution]
    style A fill:#FFE4E1
    style G fill:#90EE90
```

**Core Concepts**:

1. **Chromosome**: Solution encoded as string (e.g., binary string, real numbers)
2. **Fitness Function**: Measures quality of solution
3. **Selection**: Fitter individuals more likely to reproduce
4. **Crossover**: Combine two parents to create offspring
5. **Mutation**: Random changes to introduce diversity

**Example Applications**:
- Optimization problems (scheduling, routing)
- Parameter tuning
- Feature selection
- Neural network architecture search

````python
# Minimal Example: Genetic Algorithm for Function Optimization

def fitness_function(x):
    """Function to maximize: -(x-5)^2 + 25 (peak at x=5)"""
    return -(x - 5)**2 + 25


class GeneticAlgorithm:
    """Simple genetic algorithm."""
    
    def __init__(self, pop_size=20, mutation_rate=0.1):
        self.pop_size = pop_size
        self.mutation_rate = mutation_rate
        # Initialize population randomly in range [0, 10]
        self.population = np.random.uniform(0, 10, pop_size)
    
    def evaluate(self):
        """Evaluate fitness of all individuals."""
        return np.array([fitness_function(x) for x in self.population])
    
    def selection(self, fitness):
        """Select parents based on fitness (tournament selection)."""
        # Normalize fitness to probabilities
        fitness = fitness - fitness.min() + 1  # Make all positive
        probs = fitness / fitness.sum()
        return np.random.choice(self.population, size=2, p=probs)
    
    def crossover(self, parent1, parent2):
        """Create offspring by averaging parents."""
        alpha = np.random.random()
        return alpha * parent1 + (1 - alpha) * parent2
    
    def mutate(self, individual):
        """Add small random change."""
        if np.random.random() < self.mutation_rate:
            return individual + np.random.randn() * 0.5
        return individual
    
    def evolve(self, generations=50):
        """Run evolution for specified generations."""
        best_history = []
        avg_history = []
        
        for gen in range(generations):
            # Evaluate
            fitness = self.evaluate()
            best_history.append(fitness.max())
            avg_history.append(fitness.mean())
            
            # Create new population
            new_population = []
            for _ in range(self.pop_size):
                # Selection
                parent1, parent2 = self.selection(fitness)
                # Crossover
                offspring = self.crossover(parent1, parent2)
                # Mutation
                offspring = self.mutate(offspring)
                # Keep in bounds
                offspring = np.clip(offspring, 0, 10)
                new_population.append(offspring)
            
            self.population = np.array(new_population)
        
        return best_history, avg_history


# Run genetic algorithm
print("=== Genetic Algorithm: Function Optimization ===")
print()
print("Goal: Find x that maximizes -(x-5)^2 + 25")
print("True maximum: x = 5.0, f(5) = 25.0")
print()

ga = GeneticAlgorithm(pop_size=20, mutation_rate=0.1)
best_history, avg_history = ga.evolve(generations=50)

# Find best solution
final_fitness = ga.evaluate()
best_idx = np.argmax(final_fitness)
best_x = ga.population[best_idx]
best_f = final_fitness[best_idx]

print(f"Best solution found: x = {best_x:.4f}")
print(f"Function value: f(x) = {best_f:.4f}")
print(f"Error: {abs(5 - best_x):.4f}")

# Visualize evolution
plt.figure(figsize=(12, 5))

# Evolution progress
plt.subplot(1, 2, 1)
plt.plot(best_history, 'g-', linewidth=2, label='Best fitness')
plt.plot(avg_history, 'b--', linewidth=2, label='Average fitness')
plt.axhline(y=25, color='r', linestyle=':', label='True optimum', linewidth=2)
plt.xlabel('Generation', fontsize=12)
plt.ylabel('Fitness', fontsize=12)
plt.title('Evolution Progress', fontsize=12, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)

# Final population
plt.subplot(1, 2, 2)
x = np.linspace(0, 10, 100)
y = fitness_function(x)
plt.plot(x, y, 'b-', linewidth=2, label='Fitness function')
plt.scatter(ga.population, ga.evaluate(), c='red', s=100, 
            alpha=0.6, edgecolors='black', linewidth=1, label='Final population', zorder=5)
plt.scatter([best_x], [best_f], c='green', s=300, marker='*', 
            edgecolors='black', linewidth=2, label='Best solution', zorder=6)
plt.xlabel('x', fontsize=12)
plt.ylabel('f(x)', fontsize=12)
plt.title('Final Population Distribution', fontsize=12, fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\nNote: Solution evolved through selection, crossover, and mutation!")
````

### 1.7.2 Genetic Programming

**Genetic programming** extends genetic algorithms to evolve computer programs themselves.

**Key Difference**: Instead of evolving parameters, evolve the program structure:
- Chromosomes are programs (often represented as trees)
- Crossover swaps program subtrees
- Mutation modifies program operations

**Challenges**:
- Much more computationally expensive than genetic algorithms
- Requires enormous scale to match biological evolution
- Limited practical success compared to other AI methods

**Biological Evolution vs. Genetic Programming**:

| Aspect | Biological Evolution | Genetic Programming |
|--------|---------------------|---------------------|
| **Time** | Billions of years | Hours to days |
| **Population** | Trillions of organisms | Hundreds to thousands |
| **Computation** | Powered by the sun | Limited by computers |
| **Parallelism** | Massive parallel processing | Limited parallelism |
| **Success** | Produced all life | Modest, narrow domains |

**Future Potential**: With quantum computing and increased computational power, genetic programming may become more practical.
