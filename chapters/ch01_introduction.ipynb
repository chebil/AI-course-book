### 1.2.1 Deductive Reasoning

Deductive reasoning moves from **general principles to specific conclusions**. In AI, this means starting with a knowledge base of facts and rules, then using systematic methods (logic, search) to make inferences.

**Example:**
```
General Rule: All canine animals have four legs
Specific Fact: All dogs are canines
Deduction: Therefore, dogs have four legs
```

**Core Characteristics:**
- Starts with established hypotheses and domain knowledge
- Uses logic-based systems (propositional logic, first-order logic)
- Highly interpretable - you can explain why a conclusion was reached
- Requires extensive domain knowledge from human experts
- Provably correct within the knowledge base

**Strengths:**
- Mathematically rigorous and provably correct
- Transparent decision-making process
- Works well in well-defined, narrow domains
- Requires less data than learning-based approaches

**Limitations:**
- Combinatorial explosion in large state spaces
- Difficulty encoding intuitive, non-explicit knowledge
- Requires significant human expertise to build knowledge bases
- Brittleness - fails on cases not anticipated by knowledge base

**Algorithmic Example: Forward Chaining**
```
Algorithm: ForwardChaining(Facts, Rules)
  repeat:
    for each rule (premises → conclusion) in Rules:
      if all premises are in Facts:
        if conclusion not in Facts:
          add conclusion to Facts
  until no new facts can be inferred
```

### 1.2.2 Inductive Learning

Inductive learning moves from **specific examples to general patterns**. Instead of starting with rules, the system learns patterns from data.

**Example:**
```
Observation: I saw a couple of dogs yesterday. Both had four legs.
Induction: Therefore, all dogs have four legs.
```

Note that this is a form of potentially faulty logic based on generalizing specific experiences. However, combining it with statistical methods and large amounts of data can result in accurate (though not provably correct) inferences.

**Core Characteristics:**
- Learns from examples (data-driven)
- Discovers patterns without explicit programming
- Less interpretable ("black box" nature)
- Requires large amounts of data
- Example: Neural networks, decision trees, support vector machines

**Strengths:**
- Can capture intuitive, hard-to-articulate knowledge
- Scales with data and computational power
- Can discover novel patterns beyond human knowledge
- Works on unstructured data (images, audio, text)

**Limitations:**
- Requires massive amounts of data
- Computationally expensive to train
- Lack of interpretability (hard to explain decisions)
- Vulnerable to adversarial examples and biased data
- May learn spurious correlations

**Algorithmic Example: Decision Tree Learning**
```
Algorithm: LearnDecisionTree(Data, Labels)
  if all labels are same:
    return leaf node with that label
  
  feature, threshold ← find best split using information gain
  
  left_data ← examples where feature ≤ threshold
  right_data ← examples where feature > threshold
  
  left_tree ← LearnDecisionTree(left_data, left_labels)
  right_tree ← LearnDecisionTree(right_data, right_labels)
  
  return decision node(feature, threshold, left_tree, right_tree)
```

### 1.2.3 Comparing Deductive and Inductive Approaches

| Aspect | Deductive Reasoning | Inductive Learning |
|--------|---------------------|-----------------------|
| **Direction** | General → Specific | Specific → General |
| **Starting Point** | Rules and knowledge | Data and examples |
| **Correctness** | Provably correct | Statistically accurate |
| **Interpretability** | High (can explain reasoning) | Low (black box) |
| **Data Requirements** | Low (needs expert knowledge) | High (needs many examples) |
| **Flexibility** | Rigid, brittle | Adaptive, robust |
| **Novel Patterns** | Cannot discover | Can discover |
| **Example** | Chess engine (Stockfish) | AlphaZero |

## 1.3 Artificial General Intelligence

Most modern AI systems are **narrow** or **specialized** - they excel at specific tasks but cannot transfer their capabilities to other domains. **Artificial General Intelligence (AGI)** refers to AI systems with human-like general intelligence.

In 1950, Alan Turing proposed a test for machine intelligence that remains influential today.

**Formal Definition:**

An agent function maps from percept histories to actions:

$$f: P^* \rightarrow A$$

where \(P^*\) is the set of all possible percept sequences and \(A\) is the set of possible actions.

### PEAS Description

AI agents are characterized using **PEAS** (Performance, Environment, Actuators, Sensors):

| Agent Type | Performance | Environment | Actuators | Sensors |
|------------|-------------|-------------|-----------|---------|
| **Cleaning Robot** | Cleanliness, battery, safety | Room, furniture, dirt | Wheels, vacuum | Camera, dirt sensor |
| **Chess Program** | Win/loss/draw | Chess board, opponent | Move selection | Board position |
| **Self-driving Car** | Safety, speed, comfort | Roads, traffic | Steering, throttle | Cameras, LIDAR, GPS |
| **Medical Diagnosis** | Patient health | Patient symptoms | Treatment plan | Lab tests, imaging |

### Types of Agents

```
function REFLEX-AGENT(percept):
  if percept matches condition C1:
    return action A1
  if percept matches condition C2:
    return action A2
  ...
```

**Example**: Thermostat - if temperature < 20°C, turn on heater

**Advantages**: Simple, fast, minimal memory
**Limitations**: Cannot handle partially observable environments, no memory of past

#### 2. Model-Based Agent

Maintains an **internal state** to track aspects of the world.

**Structure:**
```
function MODEL-BASED-AGENT(percept):
  state ← UPDATE-STATE(state, action, percept, model)
  action ← CHOOSE-ACTION(state, rules)
  return action
```

**Example**: Vacuum cleaner remembering which rooms it has cleaned

**Advantages**: Handles partial observability, remembers history
**Limitations**: More complex, requires accurate world model

#### 3. Goal-Based Agent

Has explicit **goal** information and plans actions to achieve it.

**Structure:**
```
function GOAL-BASED-AGENT(percept):
  state ← UPDATE-STATE(state, action, percept, model)
  plan ← FIND-PLAN(state, goal)
  action ← FIRST-ACTION(plan)
  return action
```

**Example**: GPS navigation - planning route to destination

**Advantages**: Can plan ahead, flexible to new goals
**Limitations**: Planning can be computationally expensive

#### 4. Utility-Based Agent

Chooses actions to **maximize expected utility** (degree of happiness/success).

**Structure:**
```
function UTILITY-BASED-AGENT(percept):
  state ← UPDATE-STATE(state, action, percept, model)
  action ← argmax over actions of EXPECTED-UTILITY(action, state)
  return action
```

**Example**: Stock trading bot - maximizing profit while managing risk

**Advantages**: Can handle conflicting goals, trade-offs, uncertainty
**Limitations**: Requires utility function, more computational overhead

#### 5. Learning Agent

**Improves performance** through experience.

**Components:**
- **Performance Element**: Selects actions
- **Learning Element**: Improves performance element
- **Critic**: Provides feedback on performance
- **Problem Generator**: Suggests exploratory actions

**Example**: AlphaGo - learned to play Go through self-play

**Advantages**: Adapts to new situations, improves over time
**Limitations**: Requires training data/experience, may converge slowly

### 1.4.1 Types of Environments

The characteristics of an environment significantly affect agent design:

| Dimension | Type 1 | Type 2 | Example |
|-----------|--------|--------|---------|
| **Observability** | Fully observable | Partially observable | Chess vs. Poker |
| **Determinism** | Deterministic | Stochastic | Chess vs. Backgammon |
| **Episodes** | Episodic | Sequential | Image classification vs. Chess |
| **Dynamics** | Static | Dynamic | Crossword vs. Self-driving |
| **State Space** | Discrete | Continuous | Chess vs. Robot control |
| **Agents** | Single-agent | Multi-agent | Puzzle vs. Poker |

**Fully Observable**: Agent can see complete state (e.g., chess board)
**Partially Observable**: Agent sees only part of state (e.g., poker - can't see opponent's cards)

**Deterministic**: Next state completely determined by current state and action
**Stochastic**: Randomness affects outcomes (e.g., dice, unpredictable opponents)

**Episodic**: Each episode is independent (e.g., classifying images)
**Sequential**: Current decisions affect future (e.g., chess moves)

**Static**: Environment doesn't change while agent deliberates
**Dynamic**: Environment changes continuously (e.g., real-time strategy games)

```{note}
The simplest environment is fully observable, deterministic, episodic, static, discrete, and single-agent. The hardest is partially observable, stochastic, sequential, dynamic, continuous, and multi-agent (like real-world robotics).
```

```
Training data: (1, 3), (2, 5), (3, 7), (4, 9)

Method: Minimize sum of squared errors
  Error = Σ(actual_y - predicted_y)²

Result: y = 2x + 1

Prediction: For x=5, predict y = 2(5) + 1 = 11
```

**Analogy**: Learning with a teacher who provides correct answers

```{note}
See the [implementation notebook](ch01_introduction_implementation.ipynb) for working code examples with visualizations.
```

### 1.6.2 Unsupervised Learning

**Unsupervised learning** discovers patterns in data without labeled examples.

**Key Idea**: Find hidden structure in unlabeled data.

**Common Tasks:**

1. **Clustering**: Group similar data points
   - K-means: Partition into k clusters
   - Hierarchical: Build tree of clusters
   - DBSCAN: Density-based clustering
   - Applications: Customer segmentation, document organization

2. **Dimensionality Reduction**: Compress data to fewer dimensions
   - PCA: Principal Component Analysis
   - t-SNE: Visualizing high-dimensional data
   - Applications: Feature extraction, data visualization

3. **Anomaly Detection**: Find unusual patterns
   - Isolation Forest
   - One-class SVM
   - Applications: Fraud detection, network intrusion

**K-Means Clustering (Conceptual):**

```
Algorithm: K-Means(data, k)
  1. Initialize k random centroids
  2. Repeat until convergence:
     a. Assign each point to nearest centroid
     b. Update centroids to mean of assigned points
  3. Return cluster assignments
```

**Example:**
```
Data: Customer purchase patterns
  - Age, annual spending
  
Result: 3 clusters discovered
  - Cluster 1: Young, low spending
  - Cluster 2: Middle-aged, high spending
  - Cluster 3: Senior, medium spending
```

**Analogy**: Learning without a teacher - discovering patterns on your own

**Goal**: Maximize cumulative reward over time

$$G_t = r_t + \gamma r_{t+1} + \gamma^2 r_{t+2} + \ldots = \sum_{k=0}^{\infty} \gamma^k r_{t+k}$$

where \(\gamma \in [0,1]\) is the discount factor.

**Key Algorithms:**

1. **Value-Based**: Learn value function
   - Q-Learning
   - Deep Q-Networks (DQN)

2. **Policy-Based**: Learn policy directly
   - Policy Gradient
   - Actor-Critic

3. **Model-Based**: Learn environment model
   - AlphaZero (combines with tree search)

**Multi-Armed Bandit (Simple RL Example):**

```
Problem: Choose between k slot machines
  - Each machine has unknown reward distribution
  - Goal: Maximize total reward

Epsilon-Greedy Strategy:
  - With probability ε: explore (random action)
  - With probability 1-ε: exploit (best known action)
  
Example: 3 machines with true values [1.0, 2.0, 1.5]
  After 1000 trials:
    Learned estimates: [0.98, 2.02, 1.47]
    Machine 2 selected 70% of the time
```

**Applications:**
- Game playing: AlphaGo, AlphaZero, OpenAI Five
- Robotics: Walking, manipulation, navigation
- Autonomous vehicles: Decision making
- Resource management: Data center cooling, traffic lights

**Analogy**: Like training a dog - reward good behavior, penalize bad behavior

**Core Concepts:**

1. **Chromosome**: Solution encoded as string
   - Binary: [1,0,1,1,0,1]
   - Real-valued: [2.5, 3.1, 0.8]
   - Permutation: [3,1,4,2] for TSP

2. **Fitness Function**: Measures solution quality
   - f(chromosome) → score
   - Higher is better

3. **Selection**: Fitter individuals more likely to reproduce
   - Tournament selection
   - Roulette wheel selection
   - Rank-based selection

4. **Crossover**: Combine two parents to create offspring
   - Single-point: Split and swap
   - Two-point: Two split points
   - Uniform: Random from each parent

5. **Mutation**: Random changes for diversity
   - Bit flip for binary
   - Gaussian noise for real-valued
   - Swap for permutations

**Algorithm Outline:**

```
Algorithm: GeneticAlgorithm(population_size, generations)
  population ← random_initialization()
  
  for g = 1 to generations:
    fitness ← evaluate(population)
    
    new_population ← []
    while len(new_population) < population_size:
      parent1 ← select(population, fitness)
      parent2 ← select(population, fitness)
      
      child ← crossover(parent1, parent2)
      child ← mutate(child)
      
      add child to new_population
    
    population ← new_population
  
  return best individual from population
```

**Example Applications:**

- **Optimization**: Scheduling, routing, packing
- **Design**: Circuit design, antenna design
- **Parameter tuning**: Neural network hyperparameters
- **Feature selection**: Choosing relevant features for ML
- **Game strategy**: Evolving game-playing behaviors

```
Initial population (random trees):
  - Tree 1: (+ x 3)
  - Tree 2: (* x x)
  - Tree 3: (- (* 2 x) 1)

Crossover example:
  Parent 1: (+ (* x 2) 1)
  Parent 2: (- x 3)
  
  Swap subtrees:  
  Child: (+ (- x 3) 1)

Mutation example:
  Before: (+ x 2)
  After:  (* x 2)  (changed operator)
```

**Challenges:**

1. **Computational Expense**
   - Much more expensive than genetic algorithms
   - Programs must be executed to evaluate fitness
   - Large search space of possible programs

2. **Bloat**
   - Programs tend to grow unnecessarily large
   - Need size penalties or limits

3. **Scale**
   - Requires enormous computational resources
   - Success limited to relatively simple programs

### Biological Evolution vs. Genetic Programming

| Aspect | Biological Evolution | Genetic Programming |
|--------|---------------------|---------------------|
| **Time** | Billions of years | Hours to days |
| **Population** | Trillions of organisms | Hundreds to thousands |
| **Computation** | Powered by the sun | Limited by computers |
| **Parallelism** | Massive parallel processing | Limited parallelism |
| **Energy** | ~1-2% of solar energy | Megawatts of electricity |
| **Success** | Produced all life on Earth | Modest, narrow domains |
| **Complexity** | Human brain (86B neurons) | Simple control programs |

**Why the Gap?**

Biological evolution had:
- **Scale**: Quintillions of organisms over billions of years
- **Energy**: Nearly unlimited solar energy
- **Parallelism**: All organisms evolving simultaneously
- **Time**: 3.5 billion years to produce human intelligence

Current genetic programming has:
- Populations of thousands (not trillions)
- Days or weeks (not billions of years)
- Limited computational power

**Future Potential**:

With advances in:
- Quantum computing
- Massively parallel systems
- Neuromorphic hardware

Genetic programming may become more practical for complex problems.

## Programming Tasks

Complete implementations for all tasks are provided in [Chapter 1 - Implementation and Examples](ch01_introduction_implementation.ipynb).

### Task 1: Enhanced Reflex Agent (Easy)

Extend the vacuum agent to work in a larger environment with 4 locations arranged in a line (A-B-C-D). The agent should:
- Clean any dirty location it encounters
- Move systematically from left to right, then right to left
- Track performance (steps taken, total reward)

### Task 2: K-Means Clustering (Easy)

Implement k-means clustering from scratch and apply it to 2D synthetic data:
- Generate 3 clusters of 2D points
- Implement k-means algorithm
- Visualize the clustering result
- Compare with sklearn implementation

### Task 3: Simple Bandit Agent (Medium)

Implement and compare different reinforcement learning strategies:
- Greedy selection (always pick best)
- Epsilon-greedy (explore vs. exploit)
- UCB (Upper Confidence Bound)
- Compare their performance on a 5-armed bandit

### Task 4: Model-Based Agent (Medium)

Implement a model-based vacuum agent that:
- Maintains an internal map of which locations are clean/dirty
- Plans an optimal cleaning route
- Updates its model based on percepts
- Compare performance with reflex agent

### Task 5: Genetic Algorithm for TSP (Hard)

Use genetic algorithms to solve the Traveling Salesperson Problem:
- Create a small TSP instance (10-15 cities)
- Implement genetic algorithm with appropriate encoding
- Use order crossover to preserve valid tours
- Visualize evolution of best tour length
- Compare with random search and nearest neighbor heuristic

### Task 6: Deductive vs Inductive Comparison (Hard)

Create a medical diagnosis scenario:
- Implement rule-based (deductive) diagnostic system
- Implement decision tree (inductive) diagnostic system
- Generate synthetic patient data
- Compare: performance, interpretability, data requirements
- Use at least 5 symptoms and 3 diseases

## Summary

This chapter introduced the fundamental concepts of artificial intelligence:

### Key Concepts

1. **Two Schools of Thought**:
   - **Deductive reasoning**: Logic-based, interpretable, provably correct
   - **Inductive learning**: Data-driven, powerful, discovers patterns
   - Modern AI combines both for robust systems

2. **Agent Framework**:
   - Agents perceive and act in environments
   - PEAS: Performance, Environment, Actuators, Sensors
   - Types: Reflex, model-based, goal-based, utility-based, learning
   - Environment characteristics affect agent design

3. **Types of Learning**:
   - **Supervised**: Learn from labeled examples (classification, regression)
   - **Unsupervised**: Discover patterns in unlabeled data (clustering)
   - **Reinforcement**: Learn through trial and error with rewards

4. **Biological Inspiration**:
   - Evolution as optimization algorithm
   - Genetic algorithms for parameter optimization
   - Genetic programming for program synthesis
   - Scale challenges compared to biological evolution

### Key Takeaways

- Neither deductive nor inductive approaches alone are sufficient
- Understanding learning types guides algorithm selection
- Agent design depends on environment characteristics
- Biological principles inspire powerful AI techniques
- True AGI remains an open challenge
- Computational constraints limit evolutionary approaches

### Looking Ahead

**Chapter 2**: Search algorithms - fundamental techniques for finding solutions in state spaces
- Uninformed search (BFS, DFS, UCS)
- Informed search (A*, heuristics)
- Local search (hill climbing, simulated annealing)
- Constraint satisfaction problems

```{note}
Remember to work through the [implementation notebook](ch01_introduction_implementation.ipynb) to solidify your understanding with hands-on coding!
```
