## 1.1 Introduction

Artificial intelligence is a field that spans work from multiple communities, including classical logic programming, machine learning, and data mining. Since the founding of the field, there has been a clear dichotomy between **deductive reasoning** and **inductive learning** forms of artificial intelligence.

## 1.2 The Two Schools of Thought

### 1.2.1 Deductive Reasoning

In the deductive reasoning view, one starts with various forms of domain knowledge which are often stored as knowledge bases. These forms of domain knowledge are used to make inferences.

**Characteristics:**
- Starts with general rules and applies them to specific cases
- Highly interpretable systems
- Based on logic and systematic reasoning
- Example: Expert systems, logic programming

### 1.2.2 Inductive Learning

In learning methods, data-driven evidence is used to learn how to make predictions.

**Characteristics:**
- Starts with specific examples and generalizes to broader rules
- Data-driven approaches
- May lack interpretability
- Example: Machine learning, neural networks

````python
# Example: Simple demonstration of deductive vs inductive reasoning

# Deductive reasoning example
def deductive_reasoning():
    """
    Rule: All birds can fly
    Fact: Tweety is a bird
    Conclusion: Tweety can fly
    """
    knowledge_base = {
        'rules': ['All birds can fly'],
        'facts': ['Tweety is a bird']
    }
    
    print("Deductive Reasoning:")
    print(f"Rule: {knowledge_base['rules'][0]}")
    print(f"Fact: {knowledge_base['facts'][0]}")
    print("Conclusion: Tweety can fly")
    
deductive_reasoning()
````

## 1.3 The Concept of Agent

An **agent** is anything that can be viewed as perceiving its environment through sensors and acting upon that environment through actuators.

### Agent Components

- **Sensors**: Perceive the environment
- **Actuators**: Act on the environment
- **Agent Function**: Maps percepts to actions
