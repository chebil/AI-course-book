## 1.1 Introduction

Artificial intelligence (AI) represents humanity's attempt to create machines that can perceive, reason, learn, and act intelligently. While computers excel at computationally intensive tasks, matching human intelligence and intuition has been an aspirational goal since the inception of general-purpose computers.

The field of AI has evolved through several distinct phases:

1. **Early Excitement (1950s-1960s)**: Initial optimism about machines that could reason
2. **First AI Winter (1970s)**: Reality check due to limited computational power
3. **Expert Systems Era (1980s)**: Rise of knowledge-based systems
4. **Second AI Winter (late 1980s-1990s)**: Collapse of LISP market and funding cuts
5. **Machine Learning Renaissance (2000s-2010s)**: Revival through statistical methods
6. **Deep Learning Revolution (2010s-present)**: Breakthrough with neural networks and massive data

### What Makes Intelligence Artificial?

Intelligence in AI systems is "artificial" because it's created by humans rather than arising naturally. However, the goal is to replicate the functional capabilities of natural intelligence, even if the underlying mechanisms differ. Modern AI systems combine:

- **Deductive Reasoning**: Using logic and domain knowledge to make inferences
- **Inductive Learning**: Learning patterns from data and experience
- **Hybrid Approaches**: Combining both paradigms for robust solutions

## 1.2 The Two Schools of Thought

Since the founding of AI, there has been a fundamental dichotomy between **deductive reasoning** and **inductive learning**. Understanding this distinction is crucial for grasping the evolution and current state of AI.

### 1.2.1 Deductive Reasoning

Deductive reasoning moves from general principles to specific conclusions. In AI, this means starting with a knowledge base of facts and rules, then using systematic methods (logic, search) to make inferences.

**Core Characteristics:**
- Starts with established hypotheses and domain knowledge
- Uses logic-based systems (propositional logic, first-order logic)
- Highly interpretable - you can explain why a conclusion was reached
- Requires extensive domain knowledge from human experts
- Example: Expert systems, theorem provers, classical game-playing programs

**Historical Example: Chess Playing**

Traditional chess programs like Stockfish use deductive reasoning:
1. Human experts encode knowledge about piece values, positional factors
2. Program searches tree of possible moves up to certain depth
3. Each position is evaluated using hand-crafted evaluation function
4. Best move is selected based on minimax principle

**Strengths:**
- Provably correct within the limits of the knowledge base
- Transparent decision-making process
- Works well in well-defined, narrow domains
- Requires less data than learning-based approaches

**Limitations:**
- Combinatorial explosion in large state spaces
- Difficulty encoding intuitive, non-explicit knowledge
- Requires significant human expertise to build knowledge bases
- Brittleness - fails on cases not anticipated by knowledge base

````python
# Example: Deductive Reasoning System
# Simple forward chaining inference engine

class DeductiveSystem:
    """
    A simple forward-chaining inference engine for propositional logic.
    Demonstrates deductive reasoning from a knowledge base.
    """
    
    def __init__(self):
        self.facts = set()  # Known facts
        self.rules = []     # Inference rules (premise -> conclusion)
    
    def add_fact(self, fact):
        """Add a known fact to the knowledge base."""
        self.facts.add(fact)
        print(f"Added fact: {fact}")
    
    def add_rule(self, premises, conclusion):
        """
        Add an inference rule.
        premises: list of required facts
        conclusion: fact that can be inferred
        """
        self.rules.append((premises, conclusion))
        print(f"Added rule: {' AND '.join(premises)} -> {conclusion}")
    
    def forward_chain(self):
        """
        Apply forward chaining to derive new facts.
        Continues until no new facts can be inferred.
        """
        inferred = True
        iterations = 0
        
        while inferred:
            inferred = False
            iterations += 1
            
            for premises, conclusion in self.rules:
                # Check if all premises are satisfied
                if all(p in self.facts for p in premises):
                    # Check if conclusion is not already known
                    if conclusion not in self.facts:
                        self.facts.add(conclusion)
                        print(f"Iteration {iterations}: Inferred '{conclusion}' from {premises}")
                        inferred = True
        
        return self.facts
    
    def query(self, fact):
        """Check if a fact can be proven from the knowledge base."""
        return fact in self.facts


# Example: Animal classification system
print("=== Deductive Reasoning: Animal Classification ==")
print()

system = DeductiveSystem()

# Add facts
system.add_fact("Tweety is a bird")
system.add_fact("Polly is a bird")
system.add_fact("Polly lives in cage")
print()

# Add rules
system.add_rule(["Tweety is a bird"], "Tweety can fly")
system.add_rule(["Polly is a bird"], "Polly can fly")
system.add_rule(["Polly is a bird", "Polly lives in cage"], "Polly is a pet")
system.add_rule(["Polly is a pet"], "Polly needs care")
print()

# Perform inference
print("Performing forward chaining...")
print()
derived_facts = system.forward_chain()

print()
print("=== All Known Facts ==")
for fact in sorted(derived_facts):
    print(f"  - {fact}")

print()
print("=== Query Examples ==")
queries = ["Tweety can fly", "Polly needs care", "Tweety is a pet"]
for q in queries:
    result = "TRUE" if system.query(q) else "FALSE"
    print(f"Query: '{q}' -> {result}")
````

### 1.2.2 Inductive Learning

Inductive learning moves from specific examples to general patterns. Instead of starting with rules, the system learns patterns from data.

**Core Characteristics:**
- Learns from examples (data-driven)
- Discovers patterns without explicit programming
- Less interpretable ("black box" nature)
- Requires large amounts of data
- Example: Neural networks, decision trees, support vector machines

**Modern Example: AlphaZero Chess**

AlphaZero learns chess through self-play:
1. Starts with only the rules of chess (no domain knowledge)
2. Plays millions of games against itself
3. Neural network learns to evaluate positions from experience
4. Discovers opening theory and strategic principles independently
5. Surpasses traditional engines despite no human-encoded knowledge

**Strengths:**
- Can capture intuitive, hard-to-articulate knowledge
- Scales with data and computational power
- Can discover novel patterns beyond human knowledge
- Works on unstructured data (images, audio, text)

**Limitations:**
- Requires massive amounts of data
- Computationally expensive to train
- Lack of interpretability (hard to explain decisions)
- Vulnerable to adversarial examples and biased data
- May learn spurious correlations

````python
# Example: Inductive Learning System
# Simple decision tree learner for classification

import numpy as np
from collections import Counter

class SimpleDecisionTree:
    """
    A basic decision tree classifier demonstrating inductive learning.
    Learns patterns from training examples rather than using explicit rules.
    """
    
    def __init__(self, max_depth=3):
        self.max_depth = max_depth
        self.tree = None
    
    def entropy(self, y):
        """Calculate entropy of labels."""
        counts = Counter(y)
        probs = [count / len(y) for count in counts.values()]
        return -sum(p * np.log2(p) for p in probs if p > 0)
    
    def information_gain(self, X, y, feature_idx, threshold):
        """Calculate information gain from a split."""
        parent_entropy = self.entropy(y)
        
        # Split data
        left_mask = X[:, feature_idx] <= threshold
        right_mask = ~left_mask
        
        if sum(left_mask) == 0 or sum(right_mask) == 0:
            return 0
        
        # Calculate weighted entropy
        n = len(y)
        left_entropy = self.entropy(y[left_mask])
        right_entropy = self.entropy(y[right_mask])
        
        child_entropy = (sum(left_mask) / n * left_entropy +
                        sum(right_mask) / n * right_entropy)
        
        return parent_entropy - child_entropy
    
    def find_best_split(self, X, y):
        """Find the best feature and threshold to split on."""
        best_gain = -1
        best_feature = None
        best_threshold = None
        
        for feature_idx in range(X.shape[1]):
            thresholds = np.unique(X[:, feature_idx])
            
            for threshold in thresholds:
                gain = self.information_gain(X, y, feature_idx, threshold)
                
                if gain > best_gain:
                    best_gain = gain
                    best_feature = feature_idx
                    best_threshold = threshold
        
        return best_feature, best_threshold
    
    def build_tree(self, X, y, depth=0):
        """Recursively build the decision tree."""
        # Stopping conditions
        if depth >= self.max_depth or len(np.unique(y)) == 1:
            return Counter(y).most_common(1)[0][0]
        
        feature_idx, threshold = self.find_best_split(X, y)
        
        if feature_idx is None:
            return Counter(y).most_common(1)[0][0]
        
        # Split data
        left_mask = X[:, feature_idx] <= threshold
        right_mask = ~left_mask
        
        # Build subtrees
        left_tree = self.build_tree(X[left_mask], y[left_mask], depth + 1)
        right_tree = self.build_tree(X[right_mask], y[right_mask], depth + 1)
        
        return {
            'feature': feature_idx,
            'threshold': threshold,
            'left': left_tree,
            'right': right_tree
        }
    
    def fit(self, X, y):
        """Train the decision tree on data."""
        self.tree = self.build_tree(X, y)
        return self
    
    def predict_sample(self, x, tree):
        """Predict class for a single sample."""
        if not isinstance(tree, dict):
            return tree
        
        if x[tree['feature']] <= tree['threshold']:
            return self.predict_sample(x, tree['left'])
        else:
            return self.predict_sample(x, tree['right'])
    
    def predict(self, X):
        """Predict classes for multiple samples."""
        return np.array([self.predict_sample(x, self.tree) for x in X])


# Example: Learn to classify animals
print("=== Inductive Learning: Animal Classification ==")
print()

# Training data: [has_feathers, can_fly, lays_eggs] -> class
X_train = np.array([
    [1, 1, 1],  # Bird
    [1, 1, 1],  # Bird
    [1, 0, 1],  # Flightless bird (e.g., penguin)
    [0, 0, 0],  # Mammal
    [0, 0, 0],  # Mammal
    [0, 0, 1],  # Reptile
    [0, 0, 1],  # Reptile
    [1, 1, 1],  # Bird
])

y_train = np.array(['bird', 'bird', 'bird', 'mammal', 'mammal', 'reptile', 'reptile', 'bird'])

# Train model
tree = SimpleDecisionTree(max_depth=3)
tree.fit(X_train, y_train)
print("Model trained on", len(X_train), "examples")
print()

# Test predictions
X_test = np.array([
    [1, 1, 1],  # Should predict bird
    [0, 0, 0],  # Should predict mammal
    [1, 0, 1],  # Should predict bird (flightless)
])

predictions = tree.predict(X_test)

test_descriptions = [
    "Has feathers, can fly, lays eggs",
    "No feathers, can't fly, doesn't lay eggs",
    "Has feathers, can't fly, lays eggs"
]

print("=== Test Predictions ==")
for desc, pred in zip(test_descriptions, predictions):
    print(f"{desc} -> {pred}")

print()
print("Note: The model learned these patterns from examples, not from explicit rules!")
````

### 1.2.3 Historical Evolution and Integration

The history of AI reflects an ongoing tension between these two paradigms:

**1950s-1960s: Deductive Dominance**
- Early AI focused on logic and symbolic reasoning
- Languages like LISP (1958) and Prolog designed for deductive reasoning
- Success in toy problems ("microworlds") like Towers of Hanoi

**1970s-1980s: Expert Systems Era**
- Rule-based systems for narrow domains (medical diagnosis, mineral exploration)
- Required extensive knowledge engineering
- Brittleness and maintenance issues led to decline

**1990s-2000s: Statistical Learning Revival**
- Support Vector Machines, decision trees gain prominence
- Better performance on real-world data
- Still limited by computational power and data availability

**2010s-Present: Deep Learning Revolution**
- Massive datasets (ImageNet, Common Crawl) become available
- GPU acceleration enables training of deep networks
- Breakthroughs in computer vision, NLP, game playing
- AlphaGo (2016) and AlphaZero (2017) demonstrate power of pure learning

**The Future: Hybrid Systems**

Modern AI increasingly combines both approaches:
- Neural-symbolic AI integrates logic with learning
- Transfer learning uses pre-learned knowledge (deductive flavor) with fine-tuning (inductive)
- IBM Watson combines knowledge graphs with machine learning
- Physics-informed neural networks incorporate domain knowledge into learning

## 1.3 Artificial General Intelligence

Most modern AI systems are **narrow** or **specialized** - they excel at specific tasks but cannot transfer their capabilities to other domains. **Artificial General Intelligence (AGI)** refers to AI systems with human-like general intelligence that can:

- Learn any intellectual task a human can learn
- Transfer knowledge across domains
- Reason, plan, and solve novel problems
- Communicate naturally and understand context
- Exhibit common sense reasoning

### 1.3.1 The Turing Test

In 1950, Alan Turing proposed a test for machine intelligence that remains influential today. The **Turing Test** involves three participants:

1. **Human Judge**: Asks questions to two hidden respondents
2. **Human Respondent**: Answers questions naturally
3. **Computer Respondent**: Attempts to appear human

**Test Procedure:**
- All communication is text-based (to eliminate voice/appearance cues)
- Judge cannot see or hear the respondents
- Judge asks questions and evaluates responses
- Computer "passes" if judge cannot reliably distinguish it from the human

**Significance:**
- Replaces philosophical question "Can machines think?" with operational test
- Focuses on behavioral equivalence rather than internal mechanisms
- Sets high bar requiring natural language understanding, reasoning, knowledge

**Criticisms:**
- Anthropocentric - why should machines imitate humans?
- Doesn't test all forms of intelligence (e.g., perceptual, motor skills)
- Can potentially be "gamed" through clever programming rather than true intelligence
- Doesn't address consciousness or understanding

**Current Status:**
- No system has convincingly passed the full Turing Test
- Chatbots like ChatGPT show impressive conversational ability but have clear limitations
- Specialized systems (e.g., chess programs) exceed human performance in narrow domains
- AGI remains an aspirational goal, not a present reality

### Modern Perspectives on AGI

Contemporary AI researchers debate whether AGI is:
- **Imminent**: Some predict AGI within decades through scaling current approaches
- **Distant**: Others argue fundamental breakthroughs are needed
- **Impossible**: A few contend human-level general intelligence cannot be replicated in machines
- **Undesirable**: Concerns about safety and control of superintelligent systems

## 1.4 The Concept of Agent

The **agent** is a central abstraction in AI. An agent is anything that:
1. **Perceives** its environment through sensors
2. **Acts** upon the environment through actuators
3. Makes **decisions** based on its perceptions to achieve goals

### Agent Architecture

```
Environment
     ↑         ↓
  Sensors   Actuators
     ↑         ↓
  Percepts  Actions
     ↑         ↓
   ╔═══════════╗
   ║   Agent   ║
   ║  Function ║
   ╚═══════════╝
```

### Components of an Agent

1. **Sensors**: Gather information about the environment
   - Robot: Cameras, LIDAR, microphones, tactile sensors
   - Chess program: Board position input
   - Self-driving car: Cameras, radar, GPS, speedometer

2. **Percepts**: Information received from sensors at a given time
   - Current sensor readings
   - May include percept history

3. **Agent Function**: Maps percept sequences to actions
   - Abstract mathematical function
   - Internal decision-making process

4. **Agent Program**: Concrete implementation of agent function
   - Actual code running on physical architecture

5. **Actuators**: Mechanisms for taking actions
   - Robot: Motors, grippers, wheels
   - Chess program: Move output
   - Self-driving car: Steering, throttle, brakes

### PEAS Description

AI agents are often characterized using **PEAS** (Performance, Environment, Actuators, Sensors):

| Agent Type | Performance | Environment | Actuators | Sensors |
|------------|-------------|-------------|-----------|---------|
| Cleaning Robot | Cleanliness, battery life, safety | Room, furniture, dirt | Wheels, vacuum, brushes | Camera, dirt sensor, bump sensor |
| Chess Program | Win/loss/draw | Chess board, opponent | Move selection | Board position |
| Self-driving Car | Safety, speed, comfort, legality | Roads, traffic, weather | Steering, throttle, brake | Cameras, LIDAR, GPS, speedometer |
| Chatbot | User satisfaction, accuracy | User queries, knowledge base | Text display | Keyboard input, query parsing |

````python
# Example: Simple Agent Implementations

from abc import ABC, abstractmethod
from enum import Enum
import random

class Environment:
    """Base class for environments."""
    
    @abstractmethod
    def get_percept(self):
        """Return current percept from environment."""
        pass
    
    @abstractmethod
    def execute_action(self, action):
        """Execute action and update environment."""
        pass


class Agent(ABC):
    """Base class for all agents."""
    
    def __init__(self):
        self.percept_history = []
    
    @abstractmethod
    def choose_action(self, percept):
        """Choose action based on current percept."""
        pass
    
    def perceive(self, percept):
        """Store percept in history."""
        self.percept_history.append(percept)
        return percept


# Example 1: Vacuum Cleaner Environment
class VacuumLocation(Enum):
    A = 0
    B = 1


class VacuumAction(Enum):
    SUCK = 0
    MOVE_LEFT = 1
    MOVE_RIGHT = 2


class VacuumEnvironment(Environment):
    """Two-location vacuum world."""
    
    def __init__(self):
        self.location = VacuumLocation.A
        self.dirty = {VacuumLocation.A: True, VacuumLocation.B: True}
        self.steps = 0
    
    def get_percept(self):
        """Return (location, is_dirty) tuple."""
        return (self.location, self.dirty[self.location])
    
    def execute_action(self, action):
        """Execute action and return reward."""
        self.steps += 1
        reward = 0
        
        if action == VacuumAction.SUCK:
            if self.dirty[self.location]:
                self.dirty[self.location] = False
                reward = 10  # Reward for cleaning
        elif action == VacuumAction.MOVE_LEFT:
            self.location = VacuumLocation.A
            reward = -1  # Small penalty for moving
        elif action == VacuumAction.MOVE_RIGHT:
            self.location = VacuumLocation.B
            reward = -1
        
        return reward
    
    def is_clean(self):
        """Check if environment is fully clean."""
        return not any(self.dirty.values())


# Simple Reflex Agent
class ReflexVacuumAgent(Agent):
    """
    Simple reflex agent: acts based only on current percept.
    Rules:
    - If current location dirty -> SUCK
    - If location A -> MOVE_RIGHT
    - If location B -> MOVE_LEFT
    """
    
    def choose_action(self, percept):
        location, is_dirty = percept
        
        if is_dirty:
            return VacuumAction.SUCK
        elif location == VacuumLocation.A:
            return VacuumAction.MOVE_RIGHT
        else:
            return VacuumAction.MOVE_LEFT


# Run simulation
print("=== Vacuum Cleaner Agent Simulation ==")
print()

env = VacuumEnvironment()
agent = ReflexVacuumAgent()

total_reward = 0
max_steps = 10

print("Initial state: A is dirty, B is dirty")
print()

for step in range(max_steps):
    # Perceive
    percept = env.get_percept()
    agent.perceive(percept)
    
    # Choose action
    action = agent.choose_action(percept)
    
    # Execute
    reward = env.execute_action(action)
    total_reward += reward
    
    # Report
    loc_name = "A" if percept[0] == VacuumLocation.A else "B"
    dirty_status = "dirty" if percept[1] else "clean"
    print(f"Step {step + 1}: At {loc_name} ({dirty_status}) -> {action.name} (reward: {reward})")
    
    # Check if done
    if env.is_clean():
        print()
        print(f"Environment clean after {step + 1} steps!")
        break

print()
print(f"Total reward: {total_reward}")
````

### 1.4.1 Types of Environments

The characteristics of an environment significantly affect the appropriate agent design:

#### Fully Observable vs. Partially Observable

**Fully Observable:**
- Agent's sensors provide complete state information
- Example: Chess (entire board visible)
- Simpler to design agents

**Partially Observable:**
- Sensors provide incomplete information
- Example: Poker (opponent's cards hidden), self-driving car (limited sensor range)
- Requires maintaining belief states

#### Deterministic vs. Stochastic

**Deterministic:**
- Next state completely determined by current state and action
- Example: Chess, Sudoku
- Allows perfect planning

**Stochastic:**
- Next state involves randomness/uncertainty
- Example: Dice games, robot navigation (motor noise)
- Requires probabilistic reasoning

#### Episodic vs. Sequential

**Episodic:**
- Each action episode is independent
- Example: Image classification (each image classified independently)
- No need to look ahead

**Sequential:**
- Current decisions affect future states
- Example: Chess, robot navigation
- Requires planning and lookahead

#### Static vs. Dynamic

**Static:**
- Environment doesn't change while agent deliberates
- Example: Crossword puzzle
- Can take time to compute action

**Dynamic:**
- Environment changes during deliberation
- Example: Self-driving car, video games
- Must act quickly

#### Discrete vs. Continuous

**Discrete:**
- Finite, distinct states and actions
- Example: Chess (finite positions and moves)
- Can enumerate possibilities

**Continuous:**
- Infinite state/action spaces
- Example: Robot joint angles, car steering
- Requires continuous mathematics

#### Single-Agent vs. Multi-Agent

**Single-Agent:**
- One agent acting in environment
- Example: Puzzle solving
- Only need to model environment

**Multi-Agent:**
- Multiple agents interacting
- Competitive: Chess, poker
- Cooperative: Multi-robot warehouse
- Must model other agents' behaviors

### Environment Classification Examples

| Environment | Observable | Deterministic | Episodic | Static | Discrete | Agents |
|-------------|-----------|---------------|----------|--------|----------|--------|
| Chess | Fully | Deterministic | Sequential | Static | Discrete | Multi |
| Poker | Partial | Stochastic | Sequential | Static | Discrete | Multi |
| Self-driving | Partial | Stochastic | Sequential | Dynamic | Continuous | Multi |
| Image classification | Fully | Deterministic | Episodic | Static | Discrete | Single |
| Robot soccer | Partial | Stochastic | Sequential | Dynamic | Continuous | Multi |
