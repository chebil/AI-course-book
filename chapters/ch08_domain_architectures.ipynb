```
f_t = σ(W_f × [h_{t-1}, x_t] + b_f)  // Forget gate
i_t = σ(W_i × [h_{t-1}, x_t] + b_i)  // Input gate
o_t = σ(W_o × [h_{t-1}, x_t] + b_o)  // Output gate

C_t = f_t ⊙ C_{t-1} + i_t ⊙ tanh(W_C × [h_{t-1}, x_t] + b_C)
h_t = o_t ⊙ tanh(C_t)
```

## 8.3 Transformers

### Self-Attention

$$\text{Attention}(Q,K,V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V$$

## 8.4 Implementation

See: [ch08_domain_architectures_implementation.ipynb](ch08_domain_architectures_implementation.ipynb)

Includes:
- CNN from scratch
- LSTM implementation
- Attention mechanism
- Real-world: Image classification, text generation
