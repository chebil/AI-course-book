```
Algorithm: UNIFY(p, q)

begin
    if p = q then return {}
    else if VARIABLE?(p) then return {p/q}
    else if VARIABLE?(q) then return {q/p}
    else if COMPOUND(p) and COMPOUND(q) then
        if OP(p) ≠ OP(q) then return FAIL
        θ ← UNIFY(ARGS(p)[0], ARGS(q)[0])
        for i = 1 to LENGTH(ARGS(p)) do
            θ ← θ ∪ UNIFY(SUBST(θ,ARGS(p)[i]), SUBST(θ,ARGS(q)[i]))
        return θ
    else return FAIL
end
```

## 5.3 Implementation

See: [ch05_firstorder_implementation.ipynb](ch05_firstorder_implementation.ipynb)

Includes:
- Term and predicate classes
- Unification algorithm
- FOL inference
- Real-world: Family relations reasoner
