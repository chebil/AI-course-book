```
(subject, predicate, object)
(John, worksAt, Google)
(Google, locatedIn, California)
```

**Components:**
- Entities (nodes)
- Relations (edges)
- Attributes (properties)

## 12.2 TransE Embeddings

**Translation-based Model:**

$$\mathbf{h} + \mathbf{r} \approx \mathbf{t}$$

**Loss Function:**

$$L = \sum_{(h,r,t) \in S} \sum_{(h',r,t') \in S'} [\gamma + d(h+r, t) - d(h'+r, t')]_+$$

Where $S$ are positive triples, $S'$ are negative samples.

## 12.3 Link Prediction

```
Algorithm: LINK-PREDICTION(kg, entity, relation)

begin
    h ← embedding[entity]
    r ← embedding[relation]
    
    candidates ← []
    for each entity e in kg do
        t ← embedding[e]
        score ← -distance(h + r, t)
        candidates.append((e, score))
    
    return sort candidates by score descending
end
```

## 12.4 Implementation

See: [ch12_knowledge_graphs_implementation.ipynb](ch12_knowledge_graphs_implementation.ipynb)

Includes:
- Knowledge graph structure
- TransE embeddings
- Link prediction
- Entity resolution
- Real-world: Movie recommendation system
